# 5.1

Let's look at the variance as a function of $\alpha$

Let $g(\alpha) = \operatorname{Var}(\alpha X + (1-\alpha)Y)$.

\begin{align*}
g(\alpha) &= \operatorname{Var}(\alpha X) + \operatorname{Var}((1-\alpha)Y) + 2\operatorname{Cov}(\alpha X,(1-\alpha)Y) \\
&= \alpha^2\sigma_X^2 + (1-\alpha)^2\sigma_Y^2 + 2\alpha(1-\alpha)\sigma_{XY} \\
&= \alpha^2\sigma_X^2 + (1-2\alpha+\alpha^2)\sigma_Y^2 + (2\alpha-2\alpha^2)\sigma_{XY} \\
&= \underbrace{\left(\sigma_X^2+\sigma_Y^2-2\sigma_{XY}\right)}_{A}\alpha^2 + \underbrace{2\left(\sigma_{XY}-\sigma_Y^2\right)}_{B}\alpha + \underbrace{\sigma_Y^2}_{C}.
\end{align*}

$A = \sigma_X^2+\sigma_Y^2-2\sigma_{XY} = \operatorname{Var}(X-Y) > 0$, so $g$ is convex with a unique minimum. To find minimum, we set the derivative equal to $0$, and solve for $\alpha$:

\begin{align*}
g'(\alpha) &= 2A\alpha + B = 0 \\
&\Leftrightarrow \alpha = -\frac{B}{2A} \\
&\Leftrightarrow \alpha = \frac{-2\left(\sigma_{XY}-\sigma_Y^2\right)}{2\left(\sigma_X^2+\sigma_Y^2-2\sigma_{XY}\right)} \\
&\Leftrightarrow \alpha = \frac{\sigma_Y^2-\sigma_{XY}}{\sigma_X^2+\sigma_Y^2-2\sigma_{XY}}.
\end{align*}

Since $g''(\alpha) = 2A > 0$, this is the unique global minimum.

### (a), (b), (c), (d), (e), (f)

Let $X_1,...,X_n$ be the original data, with empirical distribution function,

\begin{align*}
    \hat{F}_n=\frac{1}{n}\sum_{i=1}^n\mathbb{I}(X_i \leq x)
\end{align*}

We then draw a bootstrap sample, using the empirical distribution s.t. $X_1^*,...,X_n^* \sim \hat{F}_n$ they will all be $i.i.d.$ The probability of the j'th observation ending up in a single draw will then be $1/n$, so a single draw not being the j'th observation will have probability $1-1/n$. Since each draw in the bootstrap is independent, we can easily generalise this, first calculating the probability that the j'th observation will be in the bootstrap sample:
\begin{align*}
    \mathbb{P}(X_j \notin \{X_1^*,...,X_n^*\})
    &=\mathbb{P}(X_j \neq X_1^*,...,X_j \neq X_n^*)\\
    &=\mathbb{P}(X_j \neq X_1^*)\cdots\mathbb{P}(X_j \neq X_n^*)\\
    &=\left(1-\frac{1}{n}\right)^n\\
    & \Leftrightarrow\\
    \mathbb{P}(X_j \in \{X_1^*,...,X_n^*\})
    &=1-\mathbb{P}(X_j \notin \{X_1^*,...,X_n^*\})=1-\left(1-\frac{1}{n}\right)^n
\end{align*}

Using the above result we can now calculate the probability for the j'th observation being in the bootstrap sample for any $n$ especially for $n\in\{5,100,10000\}$. This can be seen in the plot below, where this formula is used.

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(1, 100_001)
p = 1 - (1 - 1 / n) ** n

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(n, p, color="#1f5fa8")

for n_val in [5, 100, 10_000]:
    p_val = 1 - (1 - 1 / n_val) ** n_val
    ax.axhline(p_val, color="gray", linestyle="--", linewidth=1)
    ax.plot(n_val, p_val, "o", color="#1f5fa8")
    ax.annotate(f"n = {n_val:,}: {p_val:.4f}", xy=(n_val, p_val),
                xytext=(8, 8), textcoords="offset points", fontsize=9)

ax.set_xscale("log")
ax.set_xlabel("Sample size n")
ax.set_ylabel("Probability")
ax.set_title("The j'th observation being in the bootstrap sample.")
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
