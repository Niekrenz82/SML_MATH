We start with the fact that:

\begin{align*}
1-p(X) 
&= 1-\frac{e^{\beta_0 + \beta_1 X}}{1 + e^{\beta_0 + \beta_1 X}}\\
&= \frac{1 + e^{\beta_0 + \beta_1 X}}{1 + e^{\beta_0 + \beta_1 X}}-\frac{e^{\beta_0 + \beta_1 X}}{1 + e^{\beta_0 + \beta_1 X}}\\
&= \frac{1}{1 + e^{\beta_0 + \beta_1 X}}.
\end{align*}

With this result we easy get that,

\begin{align*}
    \frac{p(X)}{1-p(X)}=\frac{\frac{e^{\beta_0 + \beta_1 X}}{1 + e^{\beta_0 + \beta_1 X}}}{\frac{1}{1 + e^{\beta_0 + \beta_1 X}}}=e^{\beta_0 + \beta_1 X}
\end{align*}
